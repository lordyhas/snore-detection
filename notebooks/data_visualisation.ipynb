```python
def show_mel_spectrogram(path, title="Mel Spectrogram"):
    mel_spectrogram = utils.extract_mel_spectrogram_db(path)
    plt.figure(figsize=(12, 5))
    librosa.display.specshow(mel_spectrogram,
                             x_axis="time", y_axis="mel")
    plt.colorbar(format="%+2.0f dB")
    plt.title(title)
    plt.tight_layout()
    plt.tight_layout()
    plt.show()


```

```python
for ns in ns_audio_samples[:3]:
    show_mel_spectrogram(ns, title="Mel Spectrogram of non snore")

```

```python
for s in s_audio_samples[:3]:
    show_mel_spectrogram(s, title="Mel Spectrogram of snore")
```
