```python
from src.data.data_manager import AudioDataset
from torch.utils.data import DataLoader
import src.features.utils as utils
from sklearn.model_selection import train_test_split
import numpy as np
```

```python
audio_samples0 = utils.load_audio_files("../data/raw/0")
audio_samples1 = utils.load_audio_files("../data/raw/1")
X = audio_samples0 + audio_samples1
y = np.array([0] * len(audio_samples0) + [1] * len(audio_samples1))
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

train_dataset = AudioDataset(X_train, y_train, train=True)
val_dataset = AudioDataset(X_val, y_val, train=False)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=4)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=4)
```
